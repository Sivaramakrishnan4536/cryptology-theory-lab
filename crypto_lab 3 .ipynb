```python
#17
text = "CRYPTOLOGYISINTERESTING"

frequency = {}
for char in text:
    if char in frequency:
        frequency[char] += 1
    else:
        frequency[char] = 1

print("Frequencies:", frequency)

most_frequent = max(frequency, key=frequency.get)
print("Most frequently occurring character:", most_frequent)
print("Frequency:", frequency[most_frequent])
```

```python
#18
text = "CRYPTOLOGYISINTERESTING"

frequency = {}
for char in text:
    if char in frequency:
        frequency[char] += 1
    else:
        frequency[char] = 1

total = len(text)

print("Character frequency and percentage:")
for char, count in frequency.items():
    percentage = (count / total) * 100
    print(char, "->", count, "->", round(percentage, 2), "%")
```

```python
#21
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

num = int(input("Enter a number to check: "))
if is_prime(num):
    print(num, "is prime")
else:
    print(num, "is not prime")

start = int(input("Enter starting value: "))
end = int(input("Enter ending value: "))

print("Prime numbers:")
for n in range(start, end + 1):
    if is_prime(n):
        print(n, end=" ")
```

```python
#22
a = int(input("Enter first integer: "))
b = int(input("Enter second integer: "))

x, y = a, b
while y != 0:
    x, y = y, x % y

gcd = x
print("GCD:", gcd)

if gcd == 1:
    print("The numbers are coprime.")
else:
    print("The numbers are not coprime.")
```

```python
#23
b = int(input("Enter integer: "))
m = 26

inverse = -1
for x in range(1, m):
    if (b * x) % m == 1:
        inverse = x
        break

if inverse != -1:
    print("Multiplicative inverse of", b, "mod", m, "is:", inverse)
else:
    print("Multiplicative inverse does not exist.")
```
