```python

import math

p = 61
q = 53

n = p * q
phi = (p - 1) * (q - 1)

e = 17
d = pow(e, -1, phi)

print("Public Key:", (e, n))
print("Private Key:", (d, n))

message = input("Enter message: ")

# Encryption
cipher = []

for ch in message:
    c = pow(ord(ch), e, n)
    cipher.append(c)

print("Encrypted Message:", cipher)

# Decryption
plain = ""

for c in cipher:
    m = pow(c, d, n)
    plain += chr(m)

print("Decrypted Message:", plain)
```

```python

# Public parameters
p = 23
g = 5

# Private keys
a = 6
b = 15

# Public keys
A = pow(g, a, p)
B = pow(g, b, p)

print("Alice Public Key:", A)
print("Bob Public Key:", B)

# Shared secret key calculation
key_A = pow(B, a, p)
key_B = pow(A, b, p)

print("Alice Shared Key:", key_A)
print("Bob Shared Key:", key_B)

if key_A == key_B:
    print("Key Exchange Successful")
else:
    print("Key Exchange Failed")
```
