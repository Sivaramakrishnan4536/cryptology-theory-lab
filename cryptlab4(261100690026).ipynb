```python
file = open("cryptology_no.txt", "r")

uppercase = 0
lowercase = 0
digits = 0
spaces = 0
special_characters = 0

for line in file:
    for ch in line:
        if ch.isupper():
            uppercase += 1
        elif ch.islower():
            lowercase += 1
        elif ch.isdigit():
            digits += 1
        elif ch.isspace():
            spaces += 1
        else:
            special_characters += 1

file.close()

print("Number of uppercase letters:", uppercase)
print("Number of lowercase letters:", lowercase)
print("Number of digits:", digits)
print("Number of spaces:", spaces)
print("Number of special characters:", special_characters)
```

```python
# Accept 5 lines and store them in input.txt

file = open("input.txt", "w")

for i in range(5):
    line = input("Enter line " + str(i + 1) + ": ")
    file.write(line + "\n")

file.close()


# Read input.txt and count different elements

file = open("input.txt", "r")

text = file.read()

file.close()

number_of_lines = len(text.splitlines())
number_of_words = len(text.split())
number_of_alphabets = 0
number_of_digits = 0
number_of_spaces = 0
number_of_special_characters = 0

for ch in text:
    if ch.isalpha():
        number_of_alphabets += 1
    elif ch.isdigit():
        number_of_digits += 1
    elif ch == " ":
        number_of_spaces += 1
    elif ch != "\n":
        number_of_special_characters += 1


# Display the counts

print("\n--- File Statistics ---")
print("Number of lines:", number_of_lines)
print("Number of words:", number_of_words)
print("Number of alphabetic characters:", number_of_alphabets)
print("Number of digits:", number_of_digits)
print("Number of spaces:", number_of_spaces)
print("Number of special characters:", number_of_special_characters)


# Convert text to uppercase and store in processed.txt

processed_file = open("processed.txt", "w")

processed_file.write(text.upper())

processed_file.close()


# Read and display processed.txt

processed_file = open("processed.txt", "r")

processed_text = processed_file.read()

print("\n--- Contents of processed.txt ---")
print(processed_text)

processed_file.close()
```

```python
import re

file = open("security_data.txt", "r")

text = file.read()

file.close()

pattern = r'\b(?:\d{1,3}\.){3}\d{1,3}\b'

ip_addresses = re.findall(pattern, text)

print("IPv4 addresses found:")

for ip in ip_addresses:
    print(ip)
```

```python
import re

file = open("security_data.txt", "r")

text = file.read()

file.close()

pattern = r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b'

emails = re.findall(pattern, text)

print("Email addresses found:")

for email in emails:
    print(email)
```

```python
import re

file = open("security_data.txt", "r")

text = file.read()

file.close()

pattern = r'\bSEC-\d{4}-\d{4}\b'

session_ids = re.findall(pattern, text)

print("Session IDs found:")

for session_id in session_ids:
    print(session_id)
```

```python
import re

file = open("security_data.txt", "r")

text = file.read()

file.close()

pattern = r'\b[0-9a-fA-F]+\b'

hex_sequences = re.findall(pattern, text)

print("Hexadecimal sequences found:")

for hex_value in hex_sequences:
    print(hex_value)
```
