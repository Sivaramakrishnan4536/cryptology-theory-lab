```python

def encrypt(plaintext, key):
    ciphertext = ""

    for i in range(len(plaintext)):
        p = ord(plaintext[i]) - 65
        k = ord(key[i]) - 65

        c = (p + k) % 26
        ciphertext += chr(c + 65)

    return ciphertext


def decrypt(ciphertext, key):
    plaintext = ""

    for i in range(len(ciphertext)):
        c = ord(ciphertext[i]) - 65
        k = ord(key[i]) - 65

        p = (c - k) % 26
        plaintext += chr(p + 65)

    return plaintext


message = input("Enter plaintext (A-Z): ").upper()
key = input("Enter key (same length): ").upper()

if len(message) != len(key) or not message.isalpha() or not key.isalpha():
    print("Invalid input! Enter equal-length alphabetic text and key.")
else:
    cipher = encrypt(message, key)
    print("Encrypted Text:", cipher)

    plain = decrypt(cipher, key)
    print("Decrypted Text:", plain)
```

```python

def rail_fence_encrypt(text, rails):
    fence = [['' for i in range(len(text))]
             for j in range(rails)]

    row = 0
    direction = 1

    for col in range(len(text)):
        fence[row][col] = text[col]

        if row == 0:
            direction = 1
        elif row == rails - 1:
            direction = -1

        row += direction

    cipher = ""

    for i in range(rails):
        for j in range(len(text)):
            cipher += fence[i][j]

    return cipher


text = input("Enter plaintext: ").replace(" ", "").upper()
rails = int(input("Enter number of rails: "))

if rails < 2:
    print("Number of rails must be at least 2")
else:
    encrypted = rail_fence_encrypt(text, rails)
    print("Encrypted Text:", encrypted)
```
